```python
# This notebook contains code necessary to compute the leaf-contraction polynomial of a tree
# The leaf-contraction polynomial is obtained via the DNC relation, but only applying deletions and leaf-contractions
# It corresponds to all the terms in the CSF whose indexing partitions don't contain 1s
# We abbreviate leaf-contraction polynomial to "LCP"
```

```python
# Similar to CSF_helper, but only deletion and leaf-contractions
def LCP_helper(F, n, seen_list):
    partitions = Partitions(n)
    # Base case: F has already been seen
    for seen_forest in seen_list:
        if seen_forest.is_isomorphic(F):
            return seen_list[seen_forest]

    LCP = [0] * len(partitions)
    # Base case: F has no internal edges
    e = get_internal_edge(F)
    if e == None:
        p = F.connected_components_sizes()
        sign = (-1)**num_singletons(p)
        index = get_index(p, partitions)
        LCP[index] = sign
        seen_list[F.copy(immutable=True)] = LCP
        return LCP

    # Recursive case: F has not been seen and has an internal edge
    else: 
        F1 = F.copy()
        F3 = F.copy()
        LCP = np.add(LCP_helper(left_operation(F1,e),n,seen_list), LCP_helper(right_operation(F3,e),n,seen_list))
        seen_list[F.copy(immutable=True)] = LCP
        return LCP            
```

```python
# Similar to CSF_iterator, but for LCP
def LCP_iterator(n, int_edges):       
    tree_list = []
    partitions_list = Partitions(n).list()
    seen_list = {}
    
    tree_iterator = graphs.trees(n)
    
    for T in tree_iterator:
        tree_list.append(T)            
        
    L_T = len(tree_list)
    L_P = len(partitions_list)
    
    # Filter by number of internal edges
    
    if int_edges == 0:
        for i in range(L_T):
            tree_list[i] = tree_list[i].copy()

            LCP_vector = LCP_helper(tree_list[i], n, seen_list)
            l_p = get_leading_partition(LCP_vector, n)
                    
            tree_list[i].plot().show()
            print("Leading partition:", l_p)

            vector_partitions = partitions_in_tree(LCP_vector, n)
            print(coefficient_printer(LCP_vector, vector_partitions))
                
                        
    else:
        for i in range(L_T):
            if number_internal_edges(tree_list[i]) == int_edges:
                tree_list[i] = tree_list[i].copy()

                LCP_vector = LCP_helper(tree_list[i], n, seen_list)
                l_p = get_leading_partition(LCP_vector, n)
                        
                tree_list[i].plot().show()
                print("Leading partition:", l_p)

                vector_partitions = partitions_in_tree(LCP_vector, n)
                print(coefficient_printer(LCP_vector, vector_partitions))
```

```python
def LCP_tree(tree):
    vertex_list = tree.vertices()
    edge_list = tree.edges(labels=False)
    length_list = len(vertex_list)
    partitions_list = Partitions(length_list).list()
    
    partitions = Partitions(len(vertex_list))
    seen_list = {}
    
#     tree = create_tree(vertex_list, edge_list)
#     tree.plot().show()
    tree = tree.copy()
    
    LCP_vector = LCP_helper(tree, len(vertex_list), seen_list)
    l_p = get_leading_partition(LCP_vector, len(vertex_list))
        
    vector_partitions = partitions_in_tree(LCP_vector, len(vertex_list))
    print(coefficient_printer(LCP_vector, vector_partitions))
    
    return LCP_vector
```

```python
# Function that returns the LCP vector from a CSF vector
def LCP_extraction(CSFvector, n):
    LCP_vector = CSFvector
    partitions = Partitions(n).list()
    
    for i in range(len(partitions)):
        lbda = partitions[i]
        if lbda[-1]==1:
            LCP_vector[i]=0
    return LCP_vector
```
