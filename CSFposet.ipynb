```python
def poset_create_non_ones(arr, n):
    parts = list(Partitions(n))
    
    # Find the index of the last nonzero entry.
    leading_index = max(i for i, val in enumerate(arr) if val != 0)
    leading_partition = parts[leading_index]
    leading_entry = arr[leading_index]
    
    # Build a dictionary keyed by partition length.
    L = len(leading_partition)
    poset = {r: [] for r in range(1, L + 1)}
    
    # Put the leading partition at the bottom level (length L)
    poset[L].append((list(leading_partition), leading_entry))
    
    # Collect any other nonzero partitions that contain no 1's,
    # placing them in the dictionary by length.
    for i, p in enumerate(parts):
        if arr[i] != 0 and count_ones(p) == 0:
            length = len(p)
            if 1 <= length < L:
                poset[length].append((list(p), arr[i]))
    
    # Collect row strings for each level with extra separation.
    rows = []
    for r in range(1, L + 1):
        row_str = "    ".join(str(elem) for elem in poset[r])
        rows.append(row_str)
    
    # Determine the maximum row length for centering.
    max_len = max(len(row) for row in rows)
    
    # Print each row centered.
    for row in rows:
        print(row.center(max_len))
        
def poset_create_by_ones(arr, n):
    parts = list(Partitions(n))
    
    # Find the index of the last nonzero entry in arr.
    leading_index = max(i for i, val in enumerate(arr) if val != 0)
    leading_partition = parts[leading_index]
    # Use the length of the leading partition as our fixed length.
    L = len(leading_partition)
    
    # Build a dictionary keyed by the count of 1's.
    # Only consider partitions of length L with a nonzero corresponding array entry.
    poset = {}
    for i, p in enumerate(parts):
        if arr[i] != 0 and len(p) == L:
            ones = count_ones(p)
            poset.setdefault(ones, []).append((list(p), arr[i]))
    
    # Build row strings for each group.
    # We now sort the groups in descending order so that the group with the fewest ones (containing the leading partition)
    # appears at the bottom.
    rows = []
    for ones_count in sorted(poset.keys(), reverse=True):
        row_str = "    ".join(str(elem) for elem in poset[ones_count])
        rows.append(row_str)
    
    # Determine the maximum row length for centering.
    max_len = max(len(row) for row in rows) if rows else 0
    
    # Print each row centered.
    for row in rows:
        print(row.center(max_len))
```
