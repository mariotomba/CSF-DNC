# Helper functions

General utilities used by the computational and exploratory notebooks in this repository.

```python
def count_ones(values):
    """Count the entries equal to one in ``values``."""
    count = 0
    for element in values:
        if element == 1:
            count += 1
    return count


def count_non_ones(values):
    """Count the entries different from one in ``values``."""
    count = 0
    for part in values:
        if part != 1:
            count = count + 1
    return count
```

```python
import re

def parse_polynomial(polynomial):
    """Parse a printed star-basis polynomial into a coefficient dictionary."""
    terms = re.findall(r'([+-]?\d+)\*st\[(.*?)\]', polynomial)
    parsed_terms = {f"st[{vars}]": int(coef) for coef, vars in terms}
    return parsed_terms

def print_difference_table(poly1, poly2):
    """Print coefficients that differ between two parsed polynomials."""
    keys = set(poly1.keys()).union(set(poly2.keys()))
    print("| Term                     | Coefficient 1 | Coefficient 2 |")
    print("|--------------------------|---------------|---------------|")
    for key in sorted(keys):
        coef1 = poly1.get(key, 0)
        coef2 = poly2.get(key, 0)
        if coef1 != coef2:
            print(f"| {key:<24} | {coef1:<13} | {coef2:<13} |")

def compare_polynomials(polynomial1, polynomial2):
    """Parse two star-basis polynomials and print their differing coefficients."""
    poly1_terms = parse_polynomial(polynomial1)
    poly2_terms = parse_polynomial(polynomial2)
    print_difference_table(poly1_terms, poly2_terms)
```

```python
def get_keys_with_equal_values(my_dict):
    """Group dictionary keys that share a repeated value."""
    value_to_keys = {}
    for key, value in my_dict.items():
#         print(value)
        if value not in value_to_keys:
            value_to_keys[value] = []
        value_to_keys[value].append(key)
    return [keys for keys in value_to_keys.values() if len(keys) > 1]
```
