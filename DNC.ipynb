```python
import numpy as np
from sage.all import *

# Deletion on e
def left_operation(F, e):
    F.delete_edge(e)
    return F

# Contraction on e plus singleton
def middle_operation(F, e):
    F.contract_edge(e)
    F.add_vertex()
    return F

# Contraction on e plus singleton and edge
def right_operation(F, e):
    F.contract_edge(e)
    F.add_vertex()
    F.add_edge(e)
    return F
```

```python
# If tree has an internal edge, it returns one; otherwise, returns None
def get_internal_edge(F):
    for e in F.edges(labels=False):
        if F.degree(e)[0] > 1 and F.degree(e)[1] > 1:
            return e
            break
    return None

def list_internal_edges(T):
    list = []
    for e in T.edges(labels=False):
        if T.degree(e)[0] > 1 and T.degree(e)[1] > 1:
            list.append(e)
    return list

def leading_partition(T):
    internal_edges = list_internal_edges(T)
    
    for e in internal_edges:
        T.delete_edge(e)
        
    return T.connected_components_sizes()

def number_internal_edges(tree):
    count = 0
    for e in tree.edges(labels=False):
        if (tree.degree(e)[0] > 1) and (tree.degree(e)[1] > 1):
            count = count + 1
    return count

def string_sign(x):
    sign = ""
    if x>0:
        sign = sign + "+"
    else:
        sign = sign + "-"

    return sign

# Input: array p
# Output: returns the number of 1's in p
def num_singletons(p):
    num = 0
    for i in p:
        if i == 1:
            num += 1
    return num

def singleton_parity(p):
    num = num_singletons(p)
    if num % 2 == 0:
        return True
    else:
        return False

# Input: an integer partition p, the set of partitions of n
# Output: returns the index of p in Partitions(n), assuming lexicographic ordering and zero-indexing
def get_index(p, partitions):
    i = 0
    for partition in partitions:
        if partition == p:
            return int(i)
        else:
            i += 1
            
def get_leading_partition(vector, n):
    partitions = Partitions(n).list()
    leading = 0 
    count = 0
    
    while count < len(partitions):
        if vector[count] != 0:
            leading = vector[count]
            last_index = count
        count += 1

    return partitions[last_index]

def create_tree(vertex_list, edge_list):
    G = Graph()
    for i in range(0, len(vertex_list)):
        G.add_vertex()
    for edge in edge_list:
        G.add_edge(edge)
        
    # G.plot().show()
    return G

def check_one(glist):
    if glist[-1]==1:
        return True
    else:
        return False
    
def number_of_ones_leading(glist):
    count = 0
    for i in range(1,len(glist)+1):
        if glist[len(glist)-i] == 1:
            count += 1
        else:
            break
    return count


def sum_abs(glist):
    abs_sum = 0
    for number in glist:
        abs_sum = abs_sum + abs(number)
    return abs_sum   

def partitions_in_tree(csfvector, n):
    partitions = Partitions(n).list()
    vector_partition_list = []
    l_p = get_leading_partition(csfvector, n)
    
    for i in range(0, len(partitions)):
        if partitions[i] != l_p:
            vector_partition_list.append(partitions[i])
        else:
            vector_partition_list.append(l_p)
            break
    
    return vector_partition_list
    
def coefficient_printer(csfvector, partitionvector):
    string = "1*st" + str(partitionvector[0]) + " "
    
    for i in range(1, len(partitionvector)):
        if csfvector[i] != 0:
            string = string + str(string_sign(csfvector[i])) + " " + str(abs(csfvector[i]))+ "*st" + str(partitionvector[i]) + " "
    
    return string

def multiplicity(number, glist):
    count = 0
    for element in glist:
        if element == number:
            count +=1
    return count

def maximum(n, m):
    if n >= m:
        return n
    else:
        return m    

def edge_adjacency(partition, leadingpartition):
    edge_adjacency_multiset = []
    
    for element in list(dict.fromkeys(leadingpartition)):
        new_multiplicity = maximum(multiplicity(element,leadingpartition)-multiplicity(element,partition),0)
        edge_adjacency_multiset.extend([element for i in range(new_multiplicity)])
    
    return edge_adjacency_multiset, partition

def longest_in_list(list):
    max_length = 0
    for element in list:
        if len(element) > max_length:
            max_length = len(element)
    
    return max_length

def longest_no_ones(partition_list, csfvector):
    no_ones_partitions = []
    maximum_no_ones = []
    maximum_one_less_no_ones = []
    maximum_two_less_no_ones = []
    
    for i in range(0, len(partition_list)):
        if check_one(partition_list[i]) == False and csfvector[i] != 0:
            no_ones_partitions.append(partition_list[i]) 
    
    max_length = longest_in_list(no_ones_partitions)
    
    for j in range(0,len(partition_list)):
        if len(partition_list[j]) == max_length and csfvector[j] != 0 and check_one(partition_list[j]) == False:
            maximum_no_ones.append(partition_list[j])
        if len(partition_list[j]) == max_length - 1 and csfvector[j] != 0 and check_one(partition_list[j]) == False:
            maximum_one_less_no_ones.append(partition_list[j])
        if len(partition_list[j]) == max_length - 2 and csfvector[j] != 0 and check_one(partition_list[j]) == False:
            maximum_two_less_no_ones.append(partition_list[j])
    
    
    return maximum_no_ones

def support(csfvector,n):
    partition_list = Partitions(n).list()
    partition_index = []
    
    for i in range(0, len(partition_list)):
        if csfvector[i] != 0:
            partition_index.append(i)
    d = len(partition_list) - len(partition_index)
    partition_index = partition_index + [0]*d

    return np.array(partition_index)

def check_caterpillar(T):
    lp = T.longest_path()
    if lp.order() == number_internal_edges(T) + 3:
        return True
    else:
        return False
    
def substract_vectors(list1,list2):
    result = list1
    for i in range(0,len(list1)):
        result[i] = int(list1[i])-int(list2[i])
    return result

def add_vectors(list1,list2):
    result = list1
    for i in range(0,len(list1)):
        result[i] = int(list1[i])+int(list2[i])
    return result
```

```python
# Input: A forest F on n vertices, a list of forests seen so far
# Output: Returns the CSF of F as a vector of length p(n), in lexicographic order
def CSF_helper(F, n, seen_list):
    partitions = Partitions(n)
    # Base case: F has already been seen
    for seen_forest in seen_list:
        if seen_forest.is_isomorphic(F):
            return seen_list[seen_forest]

    CSF = [0] * len(partitions)
    # Base case: F has no internal edges
    e = get_internal_edge(F)
    if e == None:
        p = F.connected_components_sizes()
        sign = (-1)**num_singletons(p)
        index = get_index(p, partitions)
        CSF[index] = sign
        seen_list[F.copy(immutable=True)] = CSF
        return CSF

    # Recursive case: F has not been seen and has an internal edge
    else: 
        F1 = F.copy()
        F2 = F.copy()
        F3 = F.copy()
        CSF = np.add(CSF_helper(left_operation(F1,e),n,seen_list), CSF_helper(middle_operation(F2,e),n,seen_list))
        CSF = np.add(CSF, CSF_helper(right_operation(F3,e),n,seen_list))
        seen_list[F.copy(immutable=True)] = CSF
        return CSF
```

```python
def create_caterpillar(leading):
    G = Graph()
    internal_vertices = [0]
    
    if len(leading) == 1:
        G = G.disjoint_union(graphs.StarGraph(leading[0]-1), labels='integers')
    
    else:
        for i in range(0,len(leading)-1):
            internal_vertices.append(leading[i]+internal_vertices[i])
        # print(internal_vertices)

        for i in range(0,len(leading)):
            G = G.disjoint_union(graphs.StarGraph(leading[i]-1), labels='integers')
            

        for i in range(0, len(internal_vertices)-1):
            G.add_edge([internal_vertices[i],internal_vertices[i+1]])
        # G.plot().show()
    return G

def basis_caterpillar(leading):
    G = Graph()
    internal_vertices = [0]
    
    if len(leading) == 1:
        G = G.disjoint_union(graphs.StarGraph(leading[0]-1), labels='integers')
    if leading[1] == 1:
        return None
    else:
        for i in range(1,len(leading)-1):
            internal_vertices.append(leading[i]+internal_vertices[i-1])

        internal_vertices.append(internal_vertices[-1]+leading[-1])

        for i in range(1,len(leading)):
            G = G.disjoint_union(graphs.StarGraph(leading[i]-1), labels='integers')
            
        G = G.disjoint_union(graphs.StarGraph(leading[0]-1), labels='integers')

        for i in range(0, len(internal_vertices)-1):
            G.add_edge([internal_vertices[i],internal_vertices[i+1]])
        # G.plot().show()
    return G
    
def leading_partition_list(n):
    partitions_list = Partitions(n).list()
    leadings_list = []
    
    for p in partitions_list:
        if len(p) == 1:
            leadings_list.append(p)
        else:
            if p[1] != 1:
                leadings_list.append(p)
                
    return leadings_list

def basis_matrix(n):
    tree_list = []
    partitions_list = Partitions(n).list()
    leadings_list = leading_partition_list(n)
    
    seen_list = {}
    
    L_P = len(partitions_list)
    M = np.zeros((0, L_P))
    
    for partition in leadings_list:
        T = basis_caterpillar(partition)
        tree_list.append(T)
        
    for i in range(0,len(tree_list)):
        tree_list[i] = tree_list[i].copy()
        # tree_list[i].plot().show()
        CSF_vector = CSF_helper(tree_list[i], len(tree_list[i].vertices()), seen_list)
        # print(CSF_vector)
        M = np.vstack((M, CSF_vector))
    return M

# Input: basismatrix is the CSF matrix where CSF vectors are written in rows
# Input: CSF row vector is the CSF of the tree whose linear combination we are trying to find

def find_basis_coefficients(basismatrix, csfvector):
    np.set_printoptions(suppress=True)
    # csfvector_transpose = csfvector.tranpose() # gives the csf vector as a column vector
    basismatrix_transpose = basismatrix.transpose() # this gives the CSF basis matrix with vectors in columns
    
    basis_mult = np.matmul(basismatrix, basismatrix_transpose)
    # print(basis_mult)
    
    rhs_mult = np.matmul(basismatrix, csfvector)
    # print(rhs_mult)
    
    basis_mult_inverse = np.linalg.inv(basis_mult)
    # print(basis_mult_inverse)
    
    coefficients = np.matmul(basis_mult_inverse, rhs_mult)
    
    return coefficients
```

```python
def leading_coefficient(csfvector):
    for coeff in reversed(csfvector):
        if coeff != 0:
            return coeff
    return None

# Input: CSF vector, number of vertices
def linear_combination(tree):
    n = len(tree.vertices())
    CSF_vector = CSF_tree2(tree)
    
    leading = get_leading_partition(CSF_vector, n)
    top_length = len(leading)
    leading_coeff = leading_coefficient(CSF_vector)
    
    leading_partitions_used = []
    top_length_leading_partitions_used = []
    
    diff = CSF_vector
    
    while np.count_nonzero(diff) != 0:
        if singleton_parity(leading) == True:
            c = (-1)*leading_coeff
        else:
            c = leading_coeff
            
        # leading_partitions_used.append([leading, (-1) * c])
        
        if len(leading) == top_length:
            top_length_leading_partitions_used.append([leading, (-1) * c])
            
        diff = add_vectors(diff, c * CSF_tree2(basis_caterpillar(leading)))
        
        if np.count_nonzero(diff) != 0:
            leading = get_leading_partition(diff,n)
            leading_coeff = leading_coefficient(diff)
        else:
            # return leading_partitions_used
            return top_length_leading_partitions_used
```

```python
def sum_branch(caterpillar, k):
    sumleft=0
    sumright=0
    
    for i in range(0, k):
        sumleft = sumleft+caterpillar[i]
        sumright = sumright+caterpillar[-i-1]
    
    return sumleft, sumright
```

```python
def count_non_ones(list):
    count = 0
    for part in list:
        if part != 1:
            count = count + 1
    return count
    
def generate_caterpillars(n):
    seen = set()
    compositions = list(Compositions(n))
    caterpillar_list = []
    
    for comp in compositions:
        if comp[0] != 1 and comp[-1] != 1:
            reversal = comp[::-1]
            if tuple(reversal) not in seen:
                seen.add(tuple(comp))
                caterpillar_list.append(comp)
    
    return caterpillar_list

def generate_proper_caterpillars(n):
    seen = set()
    compos = list(Compositions(n))
    caterpillar_list = []
    proper_compos = []
    
    for comp in compos:
        if count_non_ones(comp) == len(comp):
            proper_compos.append(comp)
    
    for comp in proper_compos:
        if comp[0] != 1 and comp[-1] != 1:
            reversal = comp[::-1]
            if tuple(reversal) not in seen:
                seen.add(tuple(comp))
                caterpillar_list.append(comp)
    
    return caterpillar_list

def caterpillar_sums(comp):
    k = len(comp)-1
    sums = set()
    sumleft = 0
    sumright = 0
    
    for i in range(0, math.floor((k-1)/2)+1):
        sumleft += comp[i]
        sumright += comp[-i-1]
            
        sums.add(frozenset({sumleft, sumright}))
    
    return sums
        
```

```python
def count_ones(lst):
    count = 0
    for element in lst:
        if element == 1:
            count += 1
    return count

def poset_create_non_ones(arr, n):
    parts = list(Partitions(n))
    
    # Find the index of the last nonzero entry.
    leading_index = max(i for i, val in enumerate(arr) if val != 0)
    leading_partition = parts[leading_index]
    leading_entry = arr[leading_index]
    
    # Build a dictionary keyed by partition length.
    L = len(leading_partition)
    poset = {r: [] for r in range(1, L + 1)}
    
    # Put the leading partition at the bottom level (length L)
    poset[L].append((list(leading_partition), leading_entry))
    
    # Collect any other nonzero partitions that contain no 1's,
    # placing them in the dictionary by length.
    for i, p in enumerate(parts):
        if arr[i] != 0 and count_ones(p) == 0:
            length = len(p)
            if 1 <= length < L:
                poset[length].append((list(p), arr[i]))
    
    # Collect row strings for each level with extra separation.
    rows = []
    for r in range(1, L + 1):
        row_str = "    ".join(str(elem) for elem in poset[r])
        rows.append(row_str)
    
    # Determine the maximum row length for centering.
    max_len = max(len(row) for row in rows)
    
    # Print each row centered.
    for row in rows:
        print(row.center(max_len))
        
def poset_create_by_ones(arr, n):
    parts = list(Partitions(n))
    
    # Find the index of the last nonzero entry in arr.
    leading_index = max(i for i, val in enumerate(arr) if val != 0)
    leading_partition = parts[leading_index]
    # Use the length of the leading partition as our fixed length.
    L = len(leading_partition)
    
    # Build a dictionary keyed by the count of 1's.
    # Only consider partitions of length L with a nonzero corresponding array entry.
    poset = {}
    for i, p in enumerate(parts):
        if arr[i] != 0 and len(p) == L:
            ones = count_ones(p)
            poset.setdefault(ones, []).append((list(p), arr[i]))
    
    # Build row strings for each group.
    # We now sort the groups in descending order so that the group with the fewest ones (containing the leading partition)
    # appears at the bottom.
    rows = []
    for ones_count in sorted(poset.keys(), reverse=True):
        row_str = "    ".join(str(elem) for elem in poset[ones_count])
        rows.append(row_str)
    
    # Determine the maximum row length for centering.
    max_len = max(len(row) for row in rows) if rows else 0
    
    # Print each row centered.
    for row in rows:
        print(row.center(max_len))
```

```python
# Function that, given a CSF vector, returns an array containing the terms in the CSF 
#     corresponding to partitions of length 1 and 2 less than the leading in the form (mu, c_mu)

def mu_nu_partitions(CSFvector,n):
    partitions = Partitions(n).list()
    leading = get_leading_partition(CSFvector, n)
    mu_nu_partitions = []
    
    for i in range(len(partitions)):
        lbda = partitions[i]
        if len(lbda) == len(leading)-1 and count_ones(lbda)==0 and CSFvector[i]!=0:
            lst = [lbda, CSFvector[i]]
            tpl = tuple(lst)
            mu_nu_partitions.append(tpl)
    
    for i in range(len(partitions)):
        lbda = partitions[i]
        if len(lbda) == len(leading)-2 and count_ones(lbda)==0 and CSFvector[i]!=0:
            lst = [lbda, CSFvector[i]]
            tpl = tuple(lst)
            mu_nu_partitions.append(tpl)
    return mu_nu_partitions
```

```python
def caterpillars_leading_nu_map(n):
    map = {}
    for comp in generate_caterpillars(n):
        if count_ones(comp)==1:
            C=create_caterpillar(comp)
            C=C.copy()
            lst = list(mu_nu_partitions(CSF_tree2(C), n))
            lst.insert(0, tuple(sorted(comp, reverse=True)))
            tpl = tuple(lst)
            map[tuple(comp)]=tpl
    return map

def caterpillar_edges(comp):
    edge_map = {}
    for i in range(len(comp)-1):
        lst = sorted([comp[i], comp[i+1]], reverse=True)
        edge = tuple(lst)
        if edge in edge_map.keys():
            curr = edge_map[edge]
            edge_map[edge] = curr+1
        else:
            edge_map[edge]=1
    return edge_map
```
